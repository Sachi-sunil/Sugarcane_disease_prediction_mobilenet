```python
import os
import shutil
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import (
    Dense, GlobalAveragePooling2D, Dropout, BatchNormalization
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import (
    EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
)
from tensorflow.keras.preprocessing.image import ImageDataGenerator

from sklearn.metrics import (
    classification_report, confusion_matrix,
    roc_curve, auc, precision_score, recall_score, f1_score
)

# ─────────────────────────────────────────────
#  1. CONFIGURATION  ← only change DATASET_ROOT if needed
# ─────────────────────────────────────────────
DATASET_ROOT = "/content/data/Images"                 # root folder
TEST_DIR = "/content/data/Images/test set" # testset/healthy, testset/unhealthy

IMG_SIZE        = (224, 224)   # MobileNetV2 default input size
BATCH_SIZE      = 32
EPOCHS_FREEZE   = 10           # Phase 1: frozen base
EPOCHS_UNFREEZE = 20           # Phase 2: fine-tune
LEARNING_RATE   = 1e-4
FINE_TUNE_LR    = 1e-5
MODEL_SAVE_PATH = "best_mobilenet_sugarcane.keras"

# ─────────────────────────────────────────────
#  2. BUILD TEMP TRAIN DIR
#
#  Keras flow_from_directory() needs:
#       train_dir/class_a/   train_dir/class_b/
#
#  Your Healthy/ and Unhealthy/ folders sit directly
#  inside dataset/, so we create a lightweight
#  _train_split/ folder with symlinks — no data copy.
# ─────────────────────────────────────────────
TRAIN_DIR = os.path.join(DATASET_ROOT, "_train_split")

def build_train_dir():
    if os.path.exists(TRAIN_DIR):
        shutil.rmtree(TRAIN_DIR)
    os.makedirs(TRAIN_DIR)

    for cls in ["Healthy", "Unhealthy"]:
        src = os.path.abspath(os.path.join(DATASET_ROOT, cls))
        dst = os.path.join(TRAIN_DIR, cls.lower())   # healthy / unhealthy
        if not os.path.exists(src):
            raise FileNotFoundError(
                f"\n❌  Folder not found: {src}"
                f"\n    Make sure DATASET_ROOT points to your dataset folder."
            )
        shutil.copytree(src, dst)

    print(f"✅ Temp train dir ready: {TRAIN_DIR}")
    print(f"   Symlinks: healthy → dataset/Healthy | unhealthy → dataset/Unhealthy")

build_train_dir()

# ─────────────────────────────────────────────
#  3. DATA AUGMENTATION & GENERATORS
# ─────────────────────────────────────────────
print("\n📂 Loading dataset...")

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=30,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=True,
    brightness_range=[0.8, 1.2],
    fill_mode="nearest",
    validation_split=0.2          # 80% train, 20% val
)

test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="training",
    shuffle=True,
    seed=42
)

val_generator = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    subset="validation",
    shuffle=False,
    seed=42
)

test_generator = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode="binary",
    shuffle=False
)

class_names = list(train_generator.class_indices.keys())
print(f"✅ Classes detected  : {class_names}")
print(f"   Train samples     : {train_generator.samples}")
print(f"   Val samples       : {val_generator.samples}")
print(f"   Test samples      : {test_generator.samples}")

# ─────────────────────────────────────────────
#  4. BUILD MOBILENETV2 MODEL
# ─────────────────────────────────────────────
print("\n🧠 Building MobileNetV2 model...")

base_model = MobileNetV2(
    input_shape=(*IMG_SIZE, 3),
    include_top=False,        # remove ImageNet head
    weights="imagenet"        # use pretrained weights
)
base_model.trainable = False  # freeze base in Phase 1

# ── Custom classification head ──────────────
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = BatchNormalization()(x)
x = Dense(256, activation="relu")(x)
x = Dropout(0.4)(x)
x = Dense(128, activation="relu")(x)
x = Dropout(0.3)(x)
output = Dense(1, activation="sigmoid")(x)   # binary: 0=healthy, 1=unhealthy

model = Model(inputs=base_model.input, outputs=output)

model.compile(
    optimizer=Adam(learning_rate=LEARNING_RATE),
    loss="binary_crossentropy",
    metrics=["accuracy", tf.keras.metrics.AUC(name="auc")]
)

model.summary()

# ─────────────────────────────────────────────
#  5. CALLBACKS
# ─────────────────────────────────────────────
callbacks = [
    EarlyStopping(
        monitor="val_accuracy",
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),
    ModelCheckpoint(
        MODEL_SAVE_PATH,
        monitor="val_accuracy",
        save_best_only=True,
        verbose=1
    ),
    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=3,
        min_lr=1e-7,
        verbose=1
    )
]

# ─────────────────────────────────────────────
#  6. PHASE 1 — Train top layers only (base frozen)
# ─────────────────────────────────────────────
print("\n🔒 Phase 1: Training classification head (MobileNetV2 frozen)...")

history1 = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS_FREEZE,
    callbacks=callbacks,
    verbose=1
)

# ─────────────────────────────────────────────
#  7. PHASE 2 — Fine-tune last 50 layers of base
# ─────────────────────────────────────────────
print("\n🔓 Phase 2: Fine-tuning last 50 layers of MobileNetV2...")

base_model.trainable = True
for layer in base_model.layers[:-50]:
    layer.trainable = False        # keep early layers frozen

model.compile(
    optimizer=Adam(learning_rate=FINE_TUNE_LR),
    loss="binary_crossentropy",
    metrics=["accuracy", tf.keras.metrics.AUC(name="auc")]
)

history2 = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS_UNFREEZE,
    callbacks=callbacks,
    verbose=1
)

# ── Merge histories ──────────────────────────
history = {}
for key in history1.history:
    history[key] = history1.history[key] + history2.history[key]

phase1_end = len(history1.history["accuracy"])

# ─────────────────────────────────────────────
#  8. EVALUATE ON TEST SET
# ─────────────────────────────────────────────
print("\n📊 Evaluating on test set...")

test_loss, test_acc, _ = model.evaluate(test_generator, verbose=1)

test_generator.reset()
y_pred_prob = model.predict(test_generator, verbose=1).flatten()
y_pred      = (y_pred_prob > 0.5).astype(int)
y_true      = test_generator.classes

fpr, tpr, _ = roc_curve(y_true, y_pred_prob)
roc_auc     = auc(fpr, tpr)
prec        = precision_score(y_true, y_pred)
rec         = recall_score(y_true, y_pred)
f1          = f1_score(y_true, y_pred)

print(f"\n   Test Accuracy : {test_acc * 100:.2f}%")
print(f"   AUC           : {roc_auc:.4f}")
print(f"   Test Loss     : {test_loss:.4f}")

print("\n📋 Classification Report:")
print(classification_report(y_true, y_pred, target_names=class_names))

# ─────────────────────────────────────────────
#  9. PLOTS  (2 × 3 panel)
# ─────────────────────────────────────────────
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle("MobileNetV2 — Sugarcane Disease Prediction", fontsize=16, fontweight="bold")

epochs_range = range(1, len(history["accuracy"]) + 1)

# — Accuracy
ax = axes[0, 0]
ax.plot(epochs_range, history["accuracy"],     label="Train", color="#2196F3")
ax.plot(epochs_range, history["val_accuracy"], label="Val",   color="#FF9800")
ax.axvline(x=phase1_end, color="gray", linestyle="--", alpha=0.6, label="Fine-tune start")
ax.set_title("Accuracy"); ax.set_xlabel("Epoch"); ax.set_ylabel("Accuracy")
ax.legend(); ax.grid(alpha=0.3)

# — Loss
ax = axes[0, 1]
ax.plot(epochs_range, history["loss"],     label="Train", color="#2196F3")
ax.plot(epochs_range, history["val_loss"], label="Val",   color="#FF9800")
ax.axvline(x=phase1_end, color="gray", linestyle="--", alpha=0.6, label="Fine-tune start")
ax.set_title("Loss"); ax.set_xlabel("Epoch"); ax.set_ylabel("Loss")
ax.legend(); ax.grid(alpha=0.3)

# — AUC
ax = axes[0, 2]
ax.plot(epochs_range, history["auc"],     label="Train", color="#2196F3")
ax.plot(epochs_range, history["val_auc"], label="Val",   color="#FF9800")
ax.axvline(x=phase1_end, color="gray", linestyle="--", alpha=0.6, label="Fine-tune start")
ax.set_title("AUC"); ax.set_xlabel("Epoch"); ax.set_ylabel("AUC")
ax.legend(); ax.grid(alpha=0.3)

# — Confusion Matrix
ax = axes[1, 0]
cm = confusion_matrix(y_true, y_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=class_names, yticklabels=class_names, ax=ax)
ax.set_title("Confusion Matrix")
ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")

# — ROC Curve
ax = axes[1, 1]
ax.plot(fpr, tpr, color="#4CAF50", lw=2, label=f"AUC = {roc_auc:.4f}")
ax.plot([0, 1], [0, 1], color="gray", linestyle="--")
ax.set_title("ROC Curve")
ax.set_xlabel("False Positive Rate"); ax.set_ylabel("True Positive Rate")
ax.legend(); ax.grid(alpha=0.3)

# — Metrics bar chart
ax = axes[1, 2]
metric_labels = ["Accuracy", "AUC", "Precision", "Recall", "F1"]
metric_values = [test_acc, roc_auc, prec, rec, f1]
bars = ax.bar(metric_labels, metric_values,
              color=["#2196F3", "#4CAF50", "#FF9800", "#9C27B0", "#F44336"])
ax.set_ylim(0, 1.15)
ax.set_title("Final Metrics Summary")
for bar, val in zip(bars, metric_values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.02,
            f"{val:.3f}", ha="center", fontsize=9, fontweight="bold")
ax.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("mobilenet_results.png", dpi=150, bbox_inches="tight")
plt.show()
print("\n✅ Plot saved → mobilenet_results.png")

# ─────────────────────────────────────────────
#  10. FINAL SUMMARY  (share with team)
# ─────────────────────────────────────────────
print("\n" + "=" * 52)
print("   MOBILENETV2 — RESULTS  (share with your team)")
print("=" * 52)
print(f"   Model      : MobileNetV2")
print(f"   Accuracy   : {test_acc * 100:.2f}%")
print(f"   AUC        : {roc_auc:.4f}")
print(f"   Precision  : {prec:.4f}")
print(f"   Recall     : {rec:.4f}")
print(f"   F1 Score   : {f1:.4f}")
print("=" * 52)
print(f"\n💾 Best model saved → {MODEL_SAVE_PATH}")

# Cleanup temp symlink folder
shutil.rmtree(TRAIN_DIR)
print("🧹 Temp _train_split folder removed.")
```
